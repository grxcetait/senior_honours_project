# Notebook 1: Generating the Neutral Hydrogen Map

This notebook generates a mock neutral hydrogen (HI) 21 cm signal from theory, extracts the power spectrum, and runs 100 realisations to place error bars on the result.

## 1. Imports

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from meer21cm import MockSimulation
from meer21cm.plot import plot_map
```

## 2. Helper Utilities

```python
def save_project_plot(fig, week_number, filename):
    """
    Save a matplotlib figure into the project outputs directory.

    Parameters
    ----------
    fig : matplotlib.figure.Figure
    week_number : int
        Output goes to .../outputs/week{week_number}/
    filename : str
        Filename including extension, e.g. 'my_plot.pdf'
    """
    base_path = "/Users/gracetait/shproject/senior_honours_project/outputs"
    week_folder = f"week{week_number}"
    target_dir = os.path.join(base_path, week_folder)

    if not os.path.exists(target_dir):
        os.makedirs(target_dir)
        print(f"Created new directory: {target_dir}")

    save_path = os.path.join(target_dir, filename)
    fig.savefig(save_path, dpi=300, bbox_inches="tight")
    print(f"Successfully saved: {save_path}")
```

## 3. Simulation Setup

Create a `MockSimulation` object for the MeerKLASS 2021 survey. The `seed` parameter controls the random realisation — we use `seed=42` here for reproducibility.

```python
mock = MockSimulation(
    survey="meerklass_2021",
    band="L",
    ra_range=(334, 357),
    dec_range=(-35, -26.5),
    flat_sky=True,
    seed=42,
)
```

### Survey area

```python
plot_map(mock.W_HI, mock.wproj, have_cbar=False)
save_project_plot(plt.gcf(), week_number=2, filename="empty_hydrogen_map.pdf")
plt.show()
```

### Frequency range and cosmological redshifts

The frequency range maps directly to cosmological redshift via the 21 cm rest frequency.

```python
plt.plot(mock.nu / 1e6, mock.z_ch)
plt.xlabel("Frequency [MHz]")
plt.ylabel("Redshift")
plt.title("Frequency vs Redshift")
save_project_plot(plt.gcf(), week_number=2, filename="frequency_vs_redshift.pdf")
plt.show()
```

## 4. HI Signal Model

Define the cosmological neutral hydrogen distribution parameters.

```python
mock.omega_hi = 5e-4         # neutral hydrogen fraction
mock.mean_amp_1 = "average_hi_temp"
mock.tracer_bias_1 = 1.0     # HI bias relative to dark matter
```

Generate the mock HI data and visualise it.

> **Note:** This is purely the 21 cm signal — no telescope beam smoothing, foreground contamination, or thermal noise. Those effects are added in Notebook 2.

```python
mock.data = mock.propagate_mock_field_to_data(mock.mock_tracer_field_1)

plot_map(mock.data, mock.wproj, W=mock.W_HI, have_cbar=True, cbar_label="T [K]")
save_project_plot(plt.gcf(), week_number=2, filename="hydrogen_map.pdf")
plt.show()
```

## 5. Power Spectrum

The power spectrum $P(k_\perp, k_\parallel)$ describes how matter is distributed across different scales in the Universe.

We first compute a 2D cylindrical power spectrum and then average it into a 1D spectrum.

```python
mock.grid_data_to_field()

mock.kparabins = np.linspace(0, 1, 11)
mock.kperpbins = np.linspace(0, 0.5, 21)
power_cy, _ = mock.get_cy_power(mock.auto_power_3d_1)
```

### Cylindrical (2D) power spectrum

```python
plt.pcolormesh(mock.kperpbins, mock.kparabins, np.log10(power_cy.T))
cbar = plt.colorbar()
cbar.set_label(r"log$_{10}$ P(k$_\perp$, k$_\parallel$) [${\rm Mpc}^{3}K^2]$)")
plt.xlabel(r"k$_\perp$ [Mpc$^{-1}$]")
plt.ylabel(r"k$_\parallel$ [Mpc$^{-1}$]")
plt.title("2D Cylindrical Power Spectrum")
save_project_plot(plt.gcf(), week_number=2, filename="power_spectrum_from_hydrogen_map.pdf")
plt.show()
```

### Spherically averaged (1D) power spectrum

```python
mock.k1dbins = np.linspace(0, 1.2, 21)
power_1d, k_1d, nmodes_1d = mock.get_1d_power(mock.auto_power_3d_1)
power_model, _, _ = mock.get_1d_power(mock.auto_power_tracer_1_model)

plt.plot(k_1d, np.log10(power_1d.T), label="mock")
plt.plot(k_1d, np.log10(power_model.T), label="theory", ls="--")
plt.xlabel(r"k [Mpc$^{-1}$]")
plt.ylabel(r"log$_{10}$ P(k) [${\rm Mpc}^{3}K^2]$)")
plt.title("1D Power Spectrum — Single Realisation")
plt.legend()
save_project_plot(plt.gcf(), week_number=2, filename="1D_power_spectrum_neutral_hydrogen_map_1_realisation.pdf")
plt.show()
```

## 6. Multiple Realisations with Error Bars

A single realisation gives one possible universe. To put meaningful error bars on the power spectrum we run 100 realisations with different random seeds and take the standard deviation.

### Core function

```python
def generate_neutral_hydrogen_map(seed):
    """
    Run a single MockSimulation and return 1D power spectrum data.

    Parameters
    ----------
    seed : int
        Random seed for this realisation.

    Returns
    -------
    k_1d : array
    mock_power_1d : array   log10 of the mock power
    theory_power_1d : array log10 of the theory power
    """
    if not isinstance(seed, (int, np.integer)):
        print("Error: seed must be an integer.")
        return

    mock = MockSimulation(
        survey="meerklass_2021",
        band="L",
        ra_range=(334, 357),
        dec_range=(-35, -26.5),
        flat_sky=True,
        seed=seed,
    )

    mock.omega_hi = 5e-4
    mock.mean_amp_1 = "average_hi_temp"
    mock.tracer_bias_1 = 1.0

    mock.data = mock.propagate_mock_field_to_data(mock.mock_tracer_field_1)
    mock.grid_data_to_field()

    mock.kparabins = np.linspace(0, 1, 11)
    mock.kperpbins = np.linspace(0, 0.5, 21)
    mock.k1dbins = np.linspace(0, 1.2, 21)

    mock.get_cy_power(mock.auto_power_3d_1)          # needed to populate auto_power_3d_1
    power_1d, k_1d, _ = mock.get_1d_power(mock.auto_power_3d_1)
    power_model, _, _ = mock.get_1d_power(mock.auto_power_tracer_1_model)

    return k_1d, np.log10(power_1d.T), np.log10(power_model.T)
```

### Run 100 realisations

```python
seeds = np.arange(1, 101)

k_1d_list, mock_power_1d_list, theory_power_1d_list = [], [], []

for i, seed in enumerate(seeds):
    print(f"Running realisation {i + 1}/100 ...")
    k_1d, mock_power_1d, theory_power_1d = generate_neutral_hydrogen_map(seed)
    k_1d_list.append(k_1d)
    mock_power_1d_list.append(mock_power_1d)
    theory_power_1d_list.append(theory_power_1d)

# Statistics across realisations
k_1d_mean       = np.mean(k_1d_list, axis=0)
k_1d_std        = np.std(k_1d_list, axis=0)
mock_power_mean = np.mean(mock_power_1d_list, axis=0)
mock_power_std  = np.std(mock_power_1d_list, axis=0)
theory_power_mean = np.mean(theory_power_1d_list, axis=0)
theory_power_std  = np.std(theory_power_1d_list, axis=0)
```

### Plot with error bars

```python
plt.plot(k_1d_mean, mock_power_mean,   label="mock",   color="blue",  linewidth=1)
plt.plot(k_1d_mean, theory_power_mean, label="theory", color="red",   linewidth=1, ls="--")
plt.errorbar(k_1d_mean, mock_power_mean,   yerr=mock_power_std,   xerr=k_1d_std,
             fmt="o",    capsize=3, color="blue", markersize=1, linewidth=1)
plt.errorbar(k_1d_mean, theory_power_mean, yerr=theory_power_std, xerr=k_1d_std,
             fmt="None", capsize=3, color="red",  markersize=1, linewidth=1)

plt.xlabel(r"k [Mpc$^{-1}$]")
plt.ylabel(r"log$_{10}$ P(k) [${\rm Mpc}^{3}K^2]$)")
plt.title("1D Power Spectrum — 100 Realisations")
plt.legend()
save_project_plot(plt.gcf(), week_number=2, filename="1D_power_spectrum_neutral_hydrogen_map_100_realisations.pdf")
plt.show()
```
